```python
# Filter the DataFrame
filtered = df[(df['Team'] == 'India') & (df['Medal'] == 'Gold') & (df['Sport'] == 'Hockey')]

# Group by Year and count medals
grouped = filtered.groupby('Year').size().reset_index(name='Gold_Medals')

print(grouped)
```

```python
# Group by Year and Medal, then count
yearly_tally = df.groupby(['Year','Team', 'Medal']).size().unstack(fill_value=0)

# Add a 'Total' column
yearly_tally['Total'] = yearly_tally.sum(axis=1)

print(yearly_tally)
```
